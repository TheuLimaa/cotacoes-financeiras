```python
from src.config import API_KEY
import requests
```

```python
url = f"https://www.alphavantage.co/query?function=TIME_SERIES_DAILY&symbol=IBM&apikey={API_KEY}"
resposta = requests.get(url)
resposta.json()
```

```python
dados = resposta.json()
dados.keys()
```

```python
import pandas as pd
```

```python
serie = dados["Time Series (Daily)"]
df = pd.DataFrame.from_dict(serie, orient ="index")
df.head()
```

```python
df = df.rename(columns = {
    "1. open": "abertura",
    "2. high": "maxima",
    "3. low": "minima",
    "4. close": "fechamento",
    "5. volume": "volume"
})
```

```python
df = df.astype({
    "abertura": float,
    "maxima": float,
    "minima": float,
    "fechamento": float,
    "volume": int
})
```

```python
df.index = pd.to_datetime(df.index)
df.index.dtype
```
